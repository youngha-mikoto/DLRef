```python
# Load Fashion MNIST images and labels.
from tensorflow import keras

(train_input, train_target), (test_input, test_target) = (
    keras.datasets.fashion_mnist.load_data()
)
```

```python
# Check the training data shapes.
print(train_input.shape, train_target.shape)
```

```python
# Check the test data shapes.
print(test_input.shape, test_target.shape)
```

```python
# Display the first ten training images.
import matplotlib.pyplot as plt

fig, axs = plt.subplots(1, 10, figsize=(10, 10))
for i in range(10):
    axs[i].imshow(train_input[i], cmap="gray_r")
    axs[i].axis("off")
plt.show()
```

```python
# Show the labels for these images.
print([train_target[i] for i in range(10)])
```

```python
# Count the samples in each class.
import numpy as np

print(np.unique(train_target, return_counts=True))
```

```python
# Scale pixel values to the range [0, 1].
train_scaled = train_input / 255.0
# Flatten each 28 x 28 image into 784 features.
train_scaled = train_scaled.reshape(-1, 28 * 28)
```

```python
# Check the flattened image shape.
print(train_scaled.shape)
```

```python
# Evaluate a linear classifier with cross-validation.
from sklearn.model_selection import cross_validate
from sklearn.linear_model import SGDClassifier

sc = SGDClassifier(loss="log_loss", max_iter=5, random_state=42)
scores = cross_validate(sc, train_scaled, train_target, n_jobs=-1)
print(np.mean(scores["test_score"]))
```

```python
import tensorflow as tf
```

```python
from tensorflow import keras
```

```python
# Reserve 20% of the training data for validation.
from sklearn.model_selection import train_test_split

train_scaled, val_scaled, train_target, val_target = train_test_split(
    train_scaled, train_target, test_size=0.2, random_state=42
)
```

```python
# Check the training split shapes.
print(train_scaled.shape, train_target.shape)
```

```python
# Check the validation split shapes.
print(val_scaled.shape, val_target.shape)
```

```python
# Create an output layer with probabilities for ten classes.
dense = keras.layers.Dense(10, activation="softmax", input_shape=(784,))
```

```python
# Build a model with a single dense layer.
model = keras.Sequential(dense)
```

```python
# Use cross-entropy loss for integer class labels.
model.compile(loss="sparse_categorical_crossentropy", metrics="accuracy")
```

```python
# Inspect the first ten training labels.
print(train_target[:10])
```

```python
# Train the model for five epochs.
model.fit(train_scaled, train_target, epochs=5)
```

```python
# Measure loss and accuracy on the validation data.
model.evaluate(val_scaled, val_target)
```
